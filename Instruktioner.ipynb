# Komma igång med projektet

## 1. Skapa projektstrukturen

Skapa följande mappar i projektets huvudmapp:

- `data/` – lagrar projektets datafiler
- `src/` – innehåller projektets källkod (source code), exempelvis funktioner och klasser
- `notebooks/` – innehåller Jupyter Notebooks för analyser och rapporter
- `.venv/` – skapas automatiskt när den virtuella miljön skapas

Skapa även följande filer:

- `README.md` – beskriver projektet och hur det används
- `requirements.txt` – listar projektets installerade Python-paket och deras versioner
- `.gitignore` – anger vilka filer och mappar Git ska ignorera

## 2. Skapa en virtuell miljö

```powershell
py -m venv .venv
```

- Skapar en virtuell miljö med namnet `.venv`
- Miljön isolerar projektets Python-paket från andra projekt

### Aktivera miljön

```powershell
.\.venv\Scripts\Activate.ps1
```

- Aktiverar den virtuella miljön
- När miljön är aktiverad visas `(.venv)` i början av terminalraden

## 3. Installera projektets paket

```powershell
py -m pip install numpy pandas matplotlib seaborn scipy
```

- Installerar de Python-paket som behövs för projektet
- Paketen installeras i den aktiverade virtuella miljön

## 4. Dokumentera projektets beroenden

```powershell
python -m pip freeze > requirements.txt
```

- Listar alla installerade paket och deras versioner
- Sparar listan i `requirements.txt`
- Gör det möjligt att återskapa projektets Python-miljö senare

## 5. Skapa en .gitignore

Lägg till följande i `.gitignore`:

```text
# Ignorera virtuell Python-miljö
.venv/

# Ignorera Python-cachefiler
__pycache__/
*.pyc
```

- Hindrar Git från att ladda upp den virtuella miljön och automatiskt skapade cachefiler
- Själva `.gitignore`-filen ska däremot laddas upp till GitHub

## 6. Skapa ett repository på GitHub

- Skapa ett nytt repository på GitHub
- Kopiera repositoryts HTTPS-länk
- Öppna terminalen i projektets huvudmapp

### Initiera Git lokalt

```powershell
git init
```

- Skapar ett lokalt Git-repository i projektmappen
- Gör det möjligt att versionshantera projektet

### Koppla projektet till GitHub

```powershell
git remote add origin DIN-GITHUB-LÄNK
```

- Kopplar det lokala Git-repot till repositoryt på GitHub
- Ersätt `DIN-GITHUB-LÄNK` med länken du kopierade från GitHub

### Byt branch-namn till main

```powershell
git branch -M main
```

- Byter namn på den aktuella branchen till `main`

### Lägg till projektets filer

```powershell
git add .
```

- Förbereder projektets filer för en commit
- Filer som anges i `.gitignore` hoppas över

### Skapa den första commiten

```powershell
git commit -m "Initial commit"
```

- Sparar en första version av projektet lokalt i Git

### Ladda upp projektet till GitHub

```powershell
git push -u origin main
```

- Laddar upp projektets filer och commit till GitHub
- `-u` kopplar den lokala branchen till GitHubs branch så att framtida push-kommandon blir enklare

## 7. Uppdatera projektet framöver

När du har gjort ändringar i projektet och vill ladda upp dem till GitHub:

```powershell
git add .
git commit -m "Beskrivning av ändringen"
git push
```

- `git add .` – förbereder ändrade filer
- `git commit` – sparar ändringarna lokalt
- `git push` – laddar upp ändringarna till GitHub

## 8. Återskapa Python-miljön

Om någon annan laddar ner projektet från GitHub kan personen skapa en egen virtuell miljö och installera projektets beroenden med:

```powershell
py -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install -r requirements.txt
```

Då installeras paketen som dokumenterats i `requirements.txt`.
