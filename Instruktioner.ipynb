## För att komma igång

- Skapa först följande mappar i projektmappen:

  - `.venv` – virtuell miljö för projektets Python-paket
  - `data` – lagrar projektets datafiler
  - `src` – innehåller projektets källkod (source code)
  - `notebooks` – innehåller Jupyter-notebooks (reports)

- Skapa även följande filer:

  - `README.md` – beskriver projektet och hur det används
  - `requirements.txt` – listar projektets Python-paket och beroenden

### Skapa en virtuell miljö

```powershell
py -m venv .venv
```

- Skapar en virtuell miljö med namnet `.venv`
- Miljön isolerar projektets Python-paket från andra projekt
